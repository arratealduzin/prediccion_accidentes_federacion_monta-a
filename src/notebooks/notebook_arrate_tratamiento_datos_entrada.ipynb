```python
# Borramos duplicados basados en 'nombre','apellido1','apellido2','poblaciondomicilio'
df_fer_2024_arrate.drop_duplicates(subset=['nombre','apellido1','apellido2','poblaciondomicilio'], keep='first', inplace=True)
```

# Licencias 2025

```python
df_fer_2025_arrate = pd.read_csv("/Users/Arrate/Desktop/Licencias_2025.csv")

df_fer_2025_arrate.info()
```

```python
df_fer_2025_arrate.head(5)
```

```python
# Borrar las mismas columnas que en licencias 2024 + especificas de 2025
df_fer_2025_arrate = df_fer_2025_arrate.drop(['Sexo', 'FechaAltaSolicitud', 'Temporada',  'Complementos',
       'ImporteComplemento', 'ImporteFMM', 'ImporteSeguro',
       'ImporteFederacionEsp', 'ImporteDescuento', 'ImporteAmpliacion',
       'Cuota_Socio', 'NIFFamiliar',
       'Cuaderno19', 'Bloqueado envíos','Telefono1',
       'Email',"NIFFamiliar", 'NCuentaBancaria', 'Telefono2', 'Acceso formulario'], axis=1)
df_fer_2025_arrate.columns = df_fer_2025_arrate.columns.str.lower()

# Creamos variable edad y borrar "fechanacimiento"
df_fer_2025_arrate["fechanacimiento"] = pd.to_datetime(df_fer_2025_arrate["fechanacimiento"])
df_fer_2025_arrate["edad"] = 2025 - df_fer_2025_arrate["fechanacimiento"].dt.year
df_fer_2025_arrate = df_fer_2025_arrate.drop("fechanacimiento", axis=1)

# Crear columna federadado_2025
df_fer_2025_arrate['federado_2025'] = df_fer_2025_arrate['nombre'].notna()

# Renombrar algunas de las columnas _2025 (especificando el año) de cara a la unión de los dataset.

columnas = ['estado', 'numeroclub', 'descripcionentidad',
       'modalidad', 'relacion',
       'tipotarjeta', 'envío tarjeta', 'alta',
       'edad']
df_fer_2025_arrate.rename(columns={col: col + "_2025" for col in columnas}, inplace=True)

```

```python
# Borramos duplicados basados en 'nombre','apellido1','apellido2','poblaciondomicilio'
df_fer_2025_arrate.drop_duplicates(subset=['nif','nombre','apellido1','apellido2','poblaciondomicilio'], keep='first', inplace=True)
```

```python
df_fer_2025_arrate.columns
```

```python
df_fer_2025_arrate.info()
```

```python
# Unión de los df de licencias 2024 + 2025
# .merge() --- how='outer' 
# conserva todas las filas de ambos dataframes, y las columnas comunes aparecen una sola vez.

columnas_comunes = list(set(df_fer_2024_arrate.columns) & set(df_fer_2025_arrate.columns))
df_federados = df_fer_2024_arrate.merge(df_fer_2025_arrate, on=columnas_comunes, how='outer')
```
